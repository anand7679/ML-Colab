## Apriori

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules,fpgrowth
```

```python
item_list=[['bread','butter','jam','milk'],
           ['bread','butter','milk'],
           ['bread','juice','curd'],
           ['butter','milk','juice']]
```

```python
encoder=TransactionEncoder()
```

```python
data=encoder.fit_transform(item_list)
data
# convert to True and false in table
```

```python
col_name=encoder.columns_
col_name
```

```python
df=pd.DataFrame(data,columns=col_name)
df
```

```python
frequent_items=apriori(df,min_support=0.5,use_colnames=True)
frequent_items
# where jam and crud has been removed coz it is less then 50%
```

```python
rules=association_rules(frequent_items,metric='confidence',min_threshold=0.6)
rules=rules.loc[:,['antecedents','consequents','support','confidence']]
rules
#min_threshold=0.6--means 60% same like confidence
```

## FP-Growth
