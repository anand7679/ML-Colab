# PCA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data=load_iris()
x=data.data
y=data.target
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
plt.title("Unscaled data")
plt.show()
```

```python
# after scaling
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
plt.scatter(x_scaled[:,0],x_scaled[:,1],c=y)
plt.title("Scaled data")
plt.show()
```

```python
#we use PCA
pca=PCA(n_components=2)
# n_components=2 --> reduce the column in to 2 column
pca_data=pca.fit_transform(x_scaled)
pca_data
```

```python
df=pd.DataFrame(pca_data,columns=['pc1','pc2'])
df["target"]=y
df
```

```python
plt.scatter(df['pc1'],df['pc2'],c=df['target'])
plt.title("PCA_data")
plt.show()
```

```python
pca=PCA(n_components=0.95)
pca_data1=pca.fit_transform(x_scaled)
pca_data1
```

```python
df=pd.DataFrame(pca_data1,columns=['pc1','pc2'])
df["target"]=y
df
# BOTH ARE SAME
```

```python
plt.scatter(df['pc1'],df['pc2'],c=df['target'])
plt.title("PCA_data")
plt.show()
```

## LDA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler
```

```python
data=load_wine()
data
```

```python
x=data.data
y=data.target
x
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
plt.title("Unscaled data")
plt.show()
# the data are overlapping
```

```python
# scaling the data
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
# plt.scatter(x_scaled[:,0],x_scaled[:,1],c=y)
# plt.title("Unscaled data")
# plt.show()
```

```python
LDA=LinearDiscriminantAnalysis(n_components=2)
LDA_data=LDA.fit_transform(x_scaled,y)
LDA_data
```

```python
plt.scatter(LDA_data[:,0],LDA_data[:,1],c=y)
plt.title("LDA_data")
plt.show()
```

```python
df=pd.DataFrame(LDA_data,columns=['LD1','LD2'])
df["target"]=y
df
```
