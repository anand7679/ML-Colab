# DecisionTREE

```python
#importing
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.datasets import load_iris
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score,mean_absolute_error,r2_score
```

# DecisionTreeClassifier

```python
# load dataset
data=load_iris()
data
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df
df["species"]=data.target # we have output in target
df
```

```python
#splitting into input and output
x=df.drop('species',axis=1)
y=df['species']
```

```python
#splitting the input and output into train and test
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
modelc=DecisionTreeClassifier(max_depth=5)
modelc.fit(xtrain,ytrain)
#Max_depth=5 to cut off the tree or branch

```

```python
ypredc=modelc.predict(xtest)
ypredc
```

```python
print("trining accuracy ",modelc.score(xtrain,ytrain))
print("testing accuracy ",modelc.score(xtest,ytest))
```

```python
print("accuracy score:",accuracy_score(ytest,ypredc))
print("precision score:",precision_score(ytest,ypredc,average="weighted"))
print("recall score:",recall_score(ytest,ypredc,average="weighted"))
print("f1 score:",f1_score(ytest,ypredc,average="weighted"))
```

```python
df=pd.read_csv("https://github.com/Jeevita-R/ML_datasets/raw/refs/heads/main/House_price_prediction.csv ")
```

```python
df
```

```python
df=pd.get_dummies(df,drop_first=True,dtype=int)
df
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
scalled=StandardScaler()
x_train=scalled.fit_transform(x_train)
x_test=scalled.transform(x_test)
```

```python
ypredr=modelr.predict(x_test)
ypredr
```

```python
print("trining accuracy ",modelr.score(x_train,y_train))
print("testing accuracy ",modelr.score(x_test,y_test))
print("R2 score",r2_score(y_test, ypredr))
print("MAE",mean_absolute_error(y_test, ypredr))
```

# Random forest Classifier

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer,fetch_california_housing
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
```

```python
data=load_breast_cancer()
data
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['cancer']=data.target
df
```

```python
x=df.drop('cancer',axis=1)
y=df['cancer']
x
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
modelrc=RandomForestClassifier(n_estimators=100,max_depth=5)
modelrc.fit(xtrain,ytrain)
#n_estimators=100 for how many trees we want
```

```python
ypredc=modelrc.predict(xtest)
ypredc
```

```python
print("trining accuracy ",modelrc.score(xtrain,ytrain))
print("testing accuracy ",modelrc.score(xtest,ytest))
```

```python
confusion_matrix(ytest,ypredc)
```

```python
print("accuracy score",accuracy_score(ytest,ypredc))
print("precision score",precision_score(ytest,ypredc))
print("recall score",recall_score(ytest,ypredc))
print("F1_score ",f1_score(ytest,ypredc))
```
