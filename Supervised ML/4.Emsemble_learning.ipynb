# ADA-BOOST classifiers

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import AdaBoostClassifier,AdaBoostRegressor
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,r2_score,mean_absolute_error
from sklearn.datasets import make_classification, make_regression# creating synthatic datasets--> random datat
```

```python
# we use n_classes=2 for classification
x,y=make_classification(n_samples=1000,n_features=5,n_classes=2,random_state=42)
# n_samples=1000 -->number of row in dataset
#n_features=5 --> how many featues we want in dataset--> we have 5 different data in input column
#n_classes=2 --> how many classes in output column in y we have "0" and "1"
```

```python
x
```

```python
y
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
base_model=DecisionTreeClassifier(max_depth=1)
model=AdaBoostClassifier(estimator=base_model,n_estimators=300,random_state=42)
model.fit(xtrain,ytrain)
```

```python
ypredc=model.predict(xtest)
```

```python
print("trining efficiency",model.score(xtrain,ytrain))
print("testing efficiency",model.score(xtest,ytest))
```

```python
confusion_matrix(ytest,ypredc)
```

```python
print("precision_score:",precision_score(ytest,ypredc))
print("recall_score",recall_score(ytest,ypredc))
```

# Classifier

```python
import pandas as pd
from sklearn.datasets import load_iris
from xgboost import XGBClassifier,XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,r2_score
```

```python
data=load_iris()
data
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df
```

```python
x=df.drop('target',axis=1)
y=df.target
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
modelc=XGBClassifier(n_estimators=100,learning_rate=0.01,max_depth=5,random_state=42)
modelc.fit(xtrain,ytrain)
```

```python
ypredc=modelc.predict(xtest)
ypredc
```

```python
print("trining accuracy",modelc.score(xtrain,ytrain))
print("testing accuracy",modelc.score(xtest,ytest))
```

```python
print("accuracy score",accuracy_score(ytest,ypredc))
```
