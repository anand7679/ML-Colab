```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor
from sklearn.datasets import load_breast_cancer
```

```python
data=load_breast_cancer()
data
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

# Classifier

```python

model=RandomForestClassifier()
Parameter={'n_estimators':[100,150,200],
           'max_depth':[5,10,15]}
grid=GridSearchCV(model,Parameter,cv=5)
grid.fit(xtrain,ytrain)
```

```python
ypred=grid.predict(xtest)
ypred
```

```python
print("traing efficiency",grid.score(xtrain,ytrain))
print("testing efficiency",grid.score(xtest,ytest))
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print("accurance score:",accuracy_score(ytest,ypred))
print("Precision score",precision_score(ytest,ypred))
print("Recall score",recall_score(ytest,ypred))
```

```python
df1=pd.read_csv(" https://raw.githubusercontent.com/Jeevita-R/ML_datasets/refs/heads/main/Fetch_california_housing.csv")
df1
```

```python
df=df1.drop("Unnamed: 0", axis=1)
df
```

```python
x=df.drop('target',axis=1)
y=df.target
x
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler() # Create an instance of StandardScaler
xtrain_scaled = scaler.fit_transform(xtrain)
xtest_scaled = scaler.transform(xtest) # Correctly transform xtest
df5 = pd.DataFrame(xtrain_scaled, columns=x.columns) # Use xtrain_scaled
df5
```

```python
model=RandomForestRegressor()
Parameter={'n_estimators':[100,150,200],
           'max_depth':[5,10,15]}
grid=GridSearchCV(model,Parameter,cv=5)
grid.fit(xtrain,ytrain)
```

```python
ypredr=grid.predict(xtest)
ypredr
```

```python
print("traing efficiency",grid.score(xtrain,ytrain))
print("testing efficiency",grid.score(xtest,ytest))
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
print("mean absolute error",mean_absolute_error(ytest,ypredr))
print("r2 score: ",r2_score(ytest,ypredr))
```
