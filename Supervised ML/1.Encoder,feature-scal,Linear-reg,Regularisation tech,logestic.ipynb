Linear Regression


```python
#importing necessary lib
import pandas as pd
from sklearn.linear_model import LinearRegression,Ridge,Lasso
# sklearn--> library
# linear_model-->module
# LinearRegression --> class
# importing perfamance martics those are  r2_score,mean_absolute_error
from sklearn.metrics import r2_score,mean_absolute_error
# for splitting data set into training and testing data set we use
from sklearn.model_selection import train_test_split
```

```python
#load and printing data
# the data set is house prediction datat set
# Now we are solving regression problem
df=pd.read_csv("https://raw.githubusercontent.com/Jeevita-R/ML_datasets/refs/heads/main/House_price_prediction.csv")
df
```

```python
#what type of data  we have
df.info()
#object datatype --> str,disc,set
```

```python
df.head()
#top 5 rows
```

```python
df.tail()#last rows
```

```python
#drop innecessary colum
df=df.drop(columns=["Unnamed: 0"],axis=1)
df
#axis=1 to remove entair column
```

```python
# here we conver the object to one hot encoder
#get_dummies -->work as one-hot encodder
#drop_first=True --> it will remove the first column in the one hot encoder, and it is optional to use
# it will reduce the no. of column
df=pd.get_dummies(df,drop_first=True,dtype=int)
df
# Suburb is removed
```

```python
#slpiting the dataset into input and output
#input
x = df.drop("Price",axis=1)
x
# here except price column every other column will be assigned to variable x
# where x contain all input column
```

```python
#output
y=df["Price"]
y
# here price column will be assigned for y
```

```python
# splitting into training and testing(70%-30%)
# test_size=0.3 size of test
# random_state=42 --> training data and testing data will be stable , means no changes will happen
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
# performing Linear Regression
# traing the model
LR=LinearRegression()
LR.fit(xtrain,ytrain)
# the model is trained
```

```python
#making the prediction
#ypred--> prediction variable
ypred=LR.predict(xtest)
ypred
```

```python
# checking the efficiency of both traing and testing datat set
# to see it model is  giving the proper result
print("Training Efficiency:",LR.score(xtrain,ytrain))
print("Training Efficiency:",LR.score(xtest,ytest))
```

```python
#performance matrices
#
print("R2 score:",r2_score(ytest,ypred) )
print("R2 score:",mean_absolute_error(ytest,ypred) )
```

Ridge regression

```python
ridge=Ridge(alpha=3)
ridge.fit(xtrain,ytrain)
# alpha=3--> in ridge it will give penalty to the cost function
# in place of lamda we use alpha=3
```

```python
ypred=ridge.predict(xtest)
ypred
```

Logistic Regression


```python
#importing lib
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer #importing the breast cancer from skleart
```

```python
# reading from sklearn lib --> breast_cancer
# 1st create obj
dataset = load_breast_cancer()
dataset
#it is in dist and array formate
```

```python
#convert it into data frame
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df
#no target col
```

```python
df["cancer"]=dataset.target
# where target is loaded to df
df
```

```python
#splitting data into input and output
x=df.drop("cancer",axis=1)
y=df["cancer"]
x
```

```python
#again splitting data in train and test
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
#model creation
model=LogisticRegression()
#model creation
model.fit(xtrain,ytrain)
```

```python
#makink predictions
ypred=model.predict(xtest)
ypred
```

```python
#checking the training and testing efficiency
print("training efficiency",model.score(xtrain,ytrain))
print("training efficiency",model.score(xtest,ytest))
```

confusion matrice




```python
conf_matrx=confusion_matrix(ytest,ypred)
conf_matrx
#59 --> TP 107-->tn
```

```python
print("Accuracy score : ",accuracy_score(ytest,ypred))
print("Precision score : ",precision_score(ytest,ypred))
print("Recall score : ",recall_score(ytest,ypred))
print("F1 score : ",f1_score(ytest,ypred))

```
