```python
# built an agent thast learn to each a goal
# problem
#state: 0,1,2,3,4,5
#starts at 0
# end/goal at 5

#Action
# Action-0= left
# Action-1= Right

#reward:
# reaching the goal --> +10
# Every other step --> -1

```

```python
import numpy as np
import random
```

```python
states=6 #from 0 - 5
actions=2 # 2 action left=0, right= 1
Q=np.zeros((states,actions)) #row * column #np.zeros--> array with zero
alpha=0.1 # learning rate --> how fast or slow the model learn
gamma=0.9 # A high value
epsilon=0.2 # it is the exploration rate--> it will take  ramdom action 20%  of the time 80% it will follow the rule
episodes=500 #how much time it should play the game OR number of games the agent has to play
```

```python
for i in range(episodes): # episodes=500
  state=0
  while state!=5:
    if random.uniform(0,1) < epsilon: # random.uniform(0,1)-->generat number from 0 to 1 # less the epsilon 2
      action=random.choice([0,1]) # choice any random value from [0-1] go right or left
    else:
      action=np.argmax(Q[state]) # for ex--> Q[2]=[1.3,4.7] # argmax--> it will give the index of largest value


    next_state=state-1 if action==0 else state+1
    next_state=max(0,min(5,next_state)) #limiting the agent to move with in 0 to 5

    #setting the rewarn
    reward=10 if next_state==5 else -1 # reward 10 panalty=-1

    #update the quality/Qtable

    Q[state,action]=Q[state,action]+alpha*(reward+gamma*np.max(Q[next_state])-Q[state,action])
    # reward+gamma*np.max(Q[next_state]) --> give the target value
    state=next_state


print(Q)
# [ 6.1999259  10.        ] -> means reached the destination , after that is has learnt and after the the output will be [ 0 , 0]
```
