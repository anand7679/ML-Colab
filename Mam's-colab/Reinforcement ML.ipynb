```python
#Build an agent that learns to reach a goal
#Problem:
# states: 0,1,2,3,4,5

# state at 0
# goal at 5

#Actions: 2
# action: 0  --> left
# action: 1  --> right

#rewards:
# Reaching goal: --> +10
# Every other steps --> -1
```

```python
import numpy as np
import random

states = 6
actions = 2

Q = np.zeros((states,actions))


alpha = 0.1 #learning rate
gamma = 0.9  #discount factor
epsilon = 0.2 # exploration rate
episodes = 500 #no.of games played

for i in range(episodes):
  state=0
  while state!=5:
    if random.uniform(0,1)<epsilon:
      action = random.choice([0,1])  # makes an action
    else:
      action = np.argmax(Q[state]) # learns from qtable

    next_state = state-1 if action==0 else state+1 #agent moves left and right
    next_state = max(0,min(5,next_state)) #agent moves within 0 to 5

    reward = 10 if next_state==5 else -1  #rewards and penalty

    #updates quality value
    Q[state,action] = Q[state,action]+alpha*(reward+gamma*np.max(Q[next_state])-Q[state,action])

    state = next_state



print(Q)


```
