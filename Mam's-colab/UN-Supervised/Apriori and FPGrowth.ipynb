Apriori

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
item_list = [["bread","butter","jam","milk"],
             ["bread","butter","milk"],
             ["bread","juice","curd"],
             ["bread","milk","juice"],
             ["butter","milk","juice"]]
```

```python
encoder = TransactionEncoder()
data = encoder.fit_transform(item_list)
col_names = encoder.columns_
df = pd.DataFrame(data,columns=col_names)
df
```

```python
frequent_items = apriori(df,min_support=0.5,use_colnames=True)
frequent_items
```

```python
rules = association_rules(frequent_items,metric="confidence",min_threshold=0.6)
rules = rules.loc[:,["antecedents","consequents","support","confidence"]]
rules
```
