# Gradient Boosting Regessor

```python
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor,GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error,confusion_matrix,accuracy_score,f1_score
```

```python
df=pd.read_csv("/content/house_price_regression_dataset.csv")
df
```

```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
modelr=GradientBoostingRegressor(n_estimators=300)
modelr.fit(Xtrain,ytrain)
```

```python
ypredr=modelr.predict(Xtest)
ypredr
```

```python
print("Training efficiency: ",modelr.score(Xtrain,ytrain))
print("Testing efficiency: ",modelr.score(Xtest,ytest))
```

```python
print("R2 score",r2_score(ytest,ypredr))
print("MAE",mean_absolute_error(ytest,ypredr))
```

# Gradient Boosting Classifier

```python
df1=pd.read_csv('/content/heart_disease.csv')
df1
```

```python
df1.dropna(inplace=True)
```

```python
df1['Exercise Habits'].unique()
```

```python
df1.drop_duplicates(inplace=True)
```

```python
df1.shape
```

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
```

```python
LE=LabelEncoder()
OE=OrdinalEncoder(categories=[['Low','Medium','High']])
```

```python
# df1.columns
```

```python
for col in df1.columns:
  if df1[col].dtype==object:
    if 'High' in df1[col].unique():
      df1[col]=OE.fit_transform(df1[[col]])
    else:
      df1[col]=LE.fit_transform(df1[col])
df1
```

```python
df1.info()
```

```python
X=df1.iloc[:,:-1]
y=df1.iloc[:,-1]
```

```python
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
modelc=GradientBoostingClassifier(n_estimators=200,learning_rate=0.03)
modelc.fit(Xtrain,ytrain)
```

```python
ypredc=modelc.predict(Xtest)
ypredc
```

```python
print("Training efficiency: ",modelc.score(Xtrain,ytrain))
print("Testing efficiency: ",modelc.score(Xtest,ytest))
```
