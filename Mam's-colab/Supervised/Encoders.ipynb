```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
LE = LabelEncoder()
df = pd.DataFrame({"Gender": ["Male","Female","Male","Male","Others","Female","Male"]})
df["Gender"] = LE.fit_transform(df["Gender"])
df
```

```python
from sklearn.preprocessing import OneHotEncoder
OHE = OneHotEncoder(sparse_output=False)
df1 = pd.DataFrame({"Product": ["Laptop","Mobile","AC","Laptop","AC","Washing Machine","Mobile"]})
df1
data = OHE.fit_transform(df1[["Product"]])
names = OHE.get_feature_names_out(["Product"])
df = pd.DataFrame(data,columns = names)
df
```
