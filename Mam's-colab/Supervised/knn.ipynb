# KNeighborsClassifier

```python
#importing librarirs
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,r2_score,mean_absolute_error
from sklearn.preprocessing import StandardScaler
```

```python
data=pd.read_csv("https://gist.github.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv")
data
```

```python
X=data.iloc[:,:4]
X
X=StandardScaler().fit_transform(X)
```

```python
y=data.iloc[:,-1]
y
```

```python
#splitting the data into training and testing
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.3,random_state=42)

```

```python
knc=KNeighborsClassifier(n_neighbors=5)
knc.fit(Xtrain,ytrain)
```

```python
ypredc=knc.predict(Xtest)
ypredc
```

```python
print("Training Efficiency: ",knc.score(Xtrain,ytrain))
print("Testing Efficiency: ",knc.score(Xtest,ytest))
```

```python
print("Accuracy :",accuracy_score(ytest,ypredc))
print("precision :",precision_score(ytest,ypredc,average="weighted"))
print("recall :",recall_score(ytest,ypredc,average="weighted"))
```

# KNN Regressor

```python
df=pd.read_csv("https://github.com/Jeevita-R/ML_datasets/raw/refs/heads/main/House_price_prediction.csv")
df
```

```python
df=df.drop("Unnamed: 0",axis=1)
df
```

```python
df=pd.get_dummies(df,drop_first=True,dtype=int)
df
```

```python
#diving data into input and output
X2=df.drop("Price",axis=1)
y2=df.Price
```

```python
#dividing the data into training and testing data
X_train,X_test,y_train,y_test=train_test_split(X2,y2,test_size=0.3,random_state=42)
```

```python
knr=KNeighborsRegressor(n_neighbors=4)
knr.fit(X_train,y_train)
```

```python
ypred2=knr.predict(X_test)
ypred2
```

```python
print("Training Efficiency: ",knr.score(X_train,y_train))
print("Testing Efficiency: ",knr.score(X_test,y_test))
```

```python
print("R2 score",r2_score(y_test,ypred2))
print("MAE",mean_absolute_error(y_test,ypred2))
```
