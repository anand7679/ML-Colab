# Normalisation

```python
import pandas as pd
from sklearn.preprocessing import Normalizer
df = pd.DataFrame({"Weight": [75,85,45,67,85,88,93,56,78,84],
                   "Height": [170,178,134,156,174,187,198,183,176,138]})
norm = Normalizer()
data = norm.fit_transform(df)
df=pd.DataFrame(data,columns=df.columns)
df
```

# StandardScaler

```python
from sklearn.preprocessing import StandardScaler
std = StandardScaler()
df = pd.DataFrame({"Weight": [75,85,45,67,85,88,93,56,78,84],
                   "Height": [170,178,134,156,174,187,198,183,176,138]})
data = std.fit_transform(df)
df = pd.DataFrame(data,columns=df.columns)
df
```
