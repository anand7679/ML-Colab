# Linear Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
#loading our dataset
df= pd.read_csv("https://raw.githubusercontent.com/Jeevita-R/ML_datasets/refs/heads/main/House_price_prediction.csv")
df
```

```python
df.info()
```

```python
df = pd.get_dummies(df,drop_first=True,dtype=int)
df
```

```python
df.info()
```

```python
df=df.drop("Unnamed: 0",axis=1)
```

```python
df
```

```python
#dividing the data
X=df.drop("Price",axis=1)
y=df.Price
X
y
```

```python
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.3,random_state=2)
```

```python
Lr= LinearRegression()
```

```python
Lr.fit(Xtrain,ytrain)
```

```python
ypred=Lr.predict(Xtest)
ypred
```

```python
print("Training: ",Lr.score(Xtrain,ytrain))
print("Testing: ",Lr.score(Xtest,ytest))
```

```python
print("MAE: ",mean_absolute_error(ytest,ypred))
```

# Ridge Regression

```python
ridge = Ridge(alpha=5)
ridge.fit(Xtrain,ytrain)
```

```python
ypred = ridge.predict(Xtest)
ypred
```

```python
print("Training Eff: ",ridge.score(Xtrain,ytrain))
print("Testing Eff: ",ridge.score(Xtest,ytest))
```

# Lasso Regression

```python
lasso = Lasso(alpha=5)
```

```python
lasso.fit(Xtrain,ytrain)
```

```python
ypred = lasso.predict(Xtest)
ypred
```

```python
print("Training Eff: ",lasso.score(Xtrain,ytrain))
print("Testing Eff: ",lasso.score(Xtest,ytest))
```
